# Fold breakdowns

Window and patient counts for each cross-validation fold, read from the split files
(`fold_k.json` + `label_counts.json`). Works for any dataset listed in `SPLIT_DIRS`.

- **AF patient**: at least one AFIB window. **N patient**: N windows only.
- **Test** = the `val` key of `fold_k.json` (the held-out fold doubles as the test set).
- Counts are windows (10 s segments) that survived segmentation and the quality check.

```python
from infrastructure.breakdowns import breakdown, patient_table, fold_summary
```

## One fold
`breakdown(dataset, fold)` gives Train / Test / All rows. `patient_table(dataset, fold, subset)` lists each patient.

```python
# Breakdown of Macro View of Each Fold
breakdown('icentia_100', fold=2)
```

```python
# Breakdwon of Patients in a Train/Test Fold
patient_table('icentia_100', fold=2, subset='Test', af_only=True)
```

## All folds side by side
`fold_summary(dataset, subset)` gives one row per fold for the chosen subset ('Train' or 'Test').

```python
fold_summary('icentia_100', subset='Test')
```

```python
fold_summary('afdb', subset='Test')
```
